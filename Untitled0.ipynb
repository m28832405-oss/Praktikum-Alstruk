```python
# 1 & 2. Membuat variabel dan menampilkan nilai beserta tipe datanya
nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

print("--- Nilai dan Tipe Data Awal ---")
print("Nilai nama:", nama, "| Tipe data:", type(nama))
print("Nilai umur:", umur, "| Tipe data:", type(umur))
print("Nilai berat:", berat, "| Tipe data:", type(berat))
print("Nilai terdaftar:", terdaftar, "| Tipe data:", type(terdaftar))

# 3. Mengubah nilai umur menjadi 20
umur = 20
print("\n--- Setelah Umur Diubah ---")
print("Nilai umur yang baru:", umur, "| Tipe data:", type(umur))
# Apa yang berubah? Nilai pada variabel 'umur' berubah dari 19 menjadi 20,
# tetapi tipe datanya tetap sama yaitu integer (<class 'int'>).

# 4. Penjelasan perbedaan "19" dan 19:
# - "19" (menggunakan tanda kutip) adalah tipe data String (teks/karakter),
#   sehingga tidak bisa digunakan untuk operasi matematika secara langsung.
# - 19 (tanpa tanda kutip) adalah tipe data Integer (angka bulat),
#   sehingga dapat digunakan untuk melakukan perhitungan aritmatika.
```
